```python
import pandas as pd
proj_path = '/content/drive/MyDrive/CSE 3521/'
df = pd.read_csv(proj_path + "data/train.csv")
df["input_text"] = df["input_text"].fillna("").astype(str) 
```

```python
import pandas as pd
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

def batch_iterator(texts, batch_size=1000):
    for i in range(0, len(texts), batch_size):
        yield texts[i:i + batch_size].tolist()

tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()
trainer = BpeTrainer(
    vocab_size=30000,
    min_frequency=2,
    special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]", "[MASK]"],
)
tokenizer.train_from_iterator(
    batch_iterator(df["input_text"]),
    trainer=trainer,
    length=len(df),  # optional, enables progress bar
)
tokenizer.save(proj_path + "modern/custom_tokenizer.json")
```
